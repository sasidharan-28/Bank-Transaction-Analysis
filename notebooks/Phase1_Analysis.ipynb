## 2. Data Understanding

- **TransactionID**: Unique identifier for the transaction.
- **CustomerID**: Unique identifier for the customer.
- **CustomerDOB**: Date of birth of the customer.
- **CustGender**: Gender of the customer.
- **CustLocation**: Location of the customer.
- **CustAccountBalance**: Account balance before/after the transaction.
- **TransactionDate**: Date when the transaction occurred.
- **TransactionTime**: Time when the transaction occurred.
- **TransactionAmount (INR)**: Amount of the transaction.

**Types:**
- Identifier: TransactionID, CustomerID
- Numerical: CustAccountBalance, TransactionTime, TransactionAmount (INR)
- Categorical: CustGender, CustLocation
- Date/Time: CustomerDOB, TransactionDate, TransactionTime
- Transaction-related: TransactionID, TransactionDate, TransactionTime, TransactionAmount (INR)
- Customer-related: CustomerID, CustomerDOB, CustGender, CustLocation, CustAccountBalance

## 4. Initial EDA & 5. Phase 1 Visualizations

```
plt.figure(figsize=(8,5))
sns.histplot(df['TransactionAmount (INR)'].dropna()[:10000], bins=50, kde=True)
plt.title('Distribution of Transaction Amount (Sampled)')
plt.xlabel('Transaction Amount (INR)')
plt.ylabel('Frequency')
plt.savefig('../outputs/figures/amount_histogram.png')
plt.show()
```

```
plt.figure(figsize=(8,5))
sns.boxplot(x=df['TransactionAmount (INR)'].dropna()[:10000])
plt.title('Box Plot of Transaction Amount (Sampled)')
plt.xlabel('Transaction Amount (INR)')
plt.savefig('../outputs/figures/amount_boxplot.png')
plt.show()
```

```
plt.figure(figsize=(8,5))
top_locations = df['CustLocation'].value_counts().head(10)
sns.barplot(x=top_locations.index, y=top_locations.values)
plt.title('Top 10 Customer Locations')
plt.xlabel('Location')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.savefig('../outputs/figures/location_barchart.png')
plt.show()
```

```
plt.figure(figsize=(8,5))
customer_freq = df['CustomerID'].value_counts().head(10)
sns.barplot(x=customer_freq.index, y=customer_freq.values)
plt.title('Top 10 Customers by Transaction Frequency')
plt.xlabel('Customer ID')
plt.ylabel('Transaction Count')
plt.xticks(rotation=45)
plt.savefig('../outputs/figures/customer_freq.png')
plt.show()
```

## 6. Phase 1 Findings

1. The dataset contains a large number of transactions with numerical features heavily right-skewed (especially transaction amounts and balances).
2. Some columns contain missing values, notably `CustomerDOB`, `CustGender`, and `CustLocation`.
3. A small number of locations account for a significant portion of transactions (e.g. Mumbai, New Delhi, Bangalore).
