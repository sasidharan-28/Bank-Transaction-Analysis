# PHASE 2 - Data Preparation and EDA

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('../data/bank_transactions.csv')
```

## 1. Data Cleaning & 2. Missing Values

```
print("Missing values before cleaning:")
print(df.isnull().sum())

# Impute categorical missing values with 'Unknown'
df['CustGender'].fillna('Unknown', inplace=True)
df['CustLocation'].fillna('Unknown', inplace=True)

# Drop rows with missing crucial numerical values
num_rows_before = len(df)
df.dropna(subset=['CustAccountBalance', 'TransactionAmount (INR)'], inplace=True)
print(f"Dropped {num_rows_before - len(df)} rows due to missing numerical values.")
```

## 3. Duplicate Handling

```
print("Duplicate rows before:", df.duplicated().sum())
df.drop_duplicates(inplace=True)
print("Duplicate rows after:", df.duplicated().sum())
```

## 4. Outlier Analysis & 5. Data Standardization

```
# Standardize category names
df['CustGender'] = df['CustGender'].str.upper().str.strip()

# Create an is_outlier flag based on IQR for TransactionAmount (but we keep the data)
Q1 = df['TransactionAmount (INR)'].quantile(0.25)
Q3 = df['TransactionAmount (INR)'].quantile(0.75)
IQR = Q3 - Q1
upper_bound = Q3 + 1.5 * IQR
df['is_amount_outlier'] = df['TransactionAmount (INR)'] > upper_bound
print(f"Flagged {df['is_amount_outlier'].sum()} transactions as high-value outliers.")

df.to_csv('../data/Bank_Transactions_Cleaned.csv', index=False)
```

## 6. Exploratory Data Analysis & 7. Visualizations

```
sample_df = df.sample(10000, random_state=42) # sample for fast plotting

plt.figure(figsize=(8,5))
sns.histplot(sample_df['CustAccountBalance'], bins=50, kde=True)
plt.title('Balance Histogram')
plt.xlabel('Account Balance')
plt.ylabel('Frequency')
plt.savefig('../outputs/figures/balance_histogram.png')
plt.show()
```

```
plt.figure(figsize=(8,5))
sns.scatterplot(data=sample_df, x='CustAccountBalance', y='TransactionAmount (INR)', alpha=0.3)
plt.title('Amount vs Balance')
plt.xlabel('Account Balance')
plt.ylabel('Transaction Amount (INR)')
plt.savefig('../outputs/figures/amount_vs_balance.png')
plt.show()
```

```
plt.figure(figsize=(8,6))
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.savefig('../outputs/figures/correlation_heatmap.png')
plt.show()
```
